```python
# Setup tokenizer and dictionary size
vocab_size = 1000 
tokenizer = Tokenizer(num_words=vocab_size, oov_token="<OOV>")
tokenizer.fit_on_texts(sentences)

# Get word index mapping
word_index = tokenizer.word_index
print(f"Total vocabulary size: {len(word_index)}")

# Convert sentences to numerical sequences
sequences = tokenizer.texts_to_sequences(sentences)

# Pad sequences to ensure uniform input length
max_length = 10
padded_sequences = pad_sequences(sequences, maxlen=max_length, padding='post', truncating='post')

# Verify the tensor shape and first sample
print("\n--- Tensor Check ---")
print(f"Shape of padded data: {padded_sequences.shape}")
print(f"First processed sample: \n{padded_sequences[0]}")
```

```python
# Define Network Architecture
model = tf.keras.Sequential([
    # Word embedding layer for semantic representation
    tf.keras.layers.Embedding(input_dim=vocab_size, output_dim=16, input_length=max_length),
    
    # Global average pooling to reduce dimensionality
    tf.keras.layers.GlobalAveragePooling1D(),
    
    # Hidden dense layer for feature extraction
    tf.keras.layers.Dense(16, activation='relu'),
    
    # Output layer for binary classification (0 or 1)
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Compile model with binary crossentropy loss and Adam optimizer
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# Print architecture summary
model.summary()
```

```python
# Force build model using input shape before training
model.build(input_shape=(None, max_length))

# Print fully built model architecture
print("--- Fully Built Model Summary ---")
model.summary()
print("\n--- Starting Model Training ---")

# Train the model
history = model.fit(
    padded_sequences, 
    labels, 
    epochs=30,
    batch_size=8,
    verbose=1  # Shows epoch-by-epoch logs
)

print("\n✅ Training complete.")
```

```python
# Unseen text samples for validation
test_sentences = [
    "This is an incredible product, highly recommended!",
    "Absolutely terrible, a complete waste of my money."
]

# Process raw text into model-ready sequences
test_sequences = tokenizer.texts_to_sequences(test_sentences)
padded_test_sequences = pad_sequences(test_sequences, maxlen=max_length, padding='post', truncating='post')

# Generate raw prediction probabilities
predictions = model.predict(padded_test_sequences, verbose=0)

print("--- Inference Results ---")
for i, sentence in enumerate(test_sentences):
    score = predictions[i][0]
    # Classification logic based on 0.5 threshold
    label = "Positive" if score >= 0.5 else "Negative"
    print(f"Text: '{sentence}'")
    print(f"Predicted: {label} (Confidence Score: {score:.4f})\n")
```
